```python
# week2 data visulaization 

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_diabetes

# Load the diabetes dataset
diabetes = load_diabetes(as_frame=True)
df = diabetes.frame.copy()

# Set the visual style
sns.set_theme(style="whitegrid", palette="deep")

# Display the first five rows
df.head()
```

```python

print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

print("\nMissing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())

df.describe().round(2)
```

```python

# Visualization 1: Distribution of disease progression

plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="target",
    bins=25,
    kde=True,
    color="steelblue"
)

# Add mean and median lines
mean_value = df["target"].mean()
median_value = df["target"].median()

plt.axvline(
    mean_value,
    color="red",
    linestyle="--",
    linewidth=2,
    label=f"Mean = {mean_value:.2f}"
)

plt.axvline(
    median_value,
    color="green",
    linestyle="-.",
    linewidth=2,
    label=f"Median = {median_value:.2f}"
)

plt.title(
    "Distribution of Disease Progression",
    fontsize=16,
    fontweight="bold"
)
plt.xlabel("Disease Progression Score (Target)", fontsize=12)
plt.ylabel("Number of Patients", fontsize=12)
plt.legend()
plt.tight_layout()
plt.show()
```

```python

# Visualization 2: Annotated correlation heatmap

# Calculate correlations between all numeric columns
corr_matrix = df.corr(numeric_only=True)

plt.figure(figsize=(12, 9))

# Create the heatmap
sns.heatmap(
    corr_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    vmin=-1,
    vmax=1,
    linewidths=0.5,
    square=True,
    cbar_kws={"label": "Correlation Coefficient"}
)

plt.title(
    "Correlation Between Dataset Features and Disease Progression",
    fontsize=15,
    fontweight="bold",
    pad=18
)

plt.xticks(rotation=45, ha="right")
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

# Visualization 3: BMI vs. disease progression

plt.figure(figsize=(10, 6))

sns.regplot(
    data=df,
    x="bmi",
    y="target",
    scatter_kws={"alpha": 0.65, "s": 45},
    line_kws={"color": "red", "linewidth": 2}
)

# Calculate correlation
correlation = df["bmi"].corr(df["target"])

plt.title(
    f"BMI and Disease Progression (Correlation = {correlation:.2f})",
    fontsize=15,
    fontweight="bold"
)
plt.xlabel("BMI (Standardized Feature Value)", fontsize=12)
plt.ylabel("Disease Progression Score", fontsize=12)

plt.annotate(
    "Upward trend: higher BMI tends to be associated\nwith higher target values",
    xy=(0.02, 0.95),
    xycoords="axes fraction",
    ha="left",
    va="top",
    fontsize=10,
    bbox=dict(boxstyle="round,pad=0.4", fc="white", alpha=0.85)
)

plt.tight_layout()
plt.show()
```

```python

# Visualization 4: Compare target values across BMI groups

# Divide observations into three groups based on BMI
df["bmi_group"] = pd.qcut(
    df["bmi"],
    q=3,
    labels=["Lower BMI", "Middle BMI", "Higher BMI"]
)

plt.figure(figsize=(9, 6))

sns.boxplot(
    data=df,
    x="bmi_group",
    y="target",
    hue="bmi_group",
    palette="Set2",
    legend=False
)

plt.title(
    "Disease Progression Across BMI Groups",
    fontsize=15,
    fontweight="bold"
)
plt.xlabel("BMI Group", fontsize=12)
plt.ylabel("Disease Progression Score", fontsize=12)

plt.tight_layout()
plt.show()

# Display group summaries
df.groupby("bmi_group", observed=True)["target"].agg(
    ["count", "median", "mean"]
).round(2)
```

```python

# Visualization 5: Rank features by correlation with the target

# Calculate correlation of each feature with the target
target_corr = df.drop(columns=["bmi_group"], errors="ignore").corr(
    numeric_only=True
)["target"].drop("target")

# Sort from strongest positive to strongest negative
target_corr = target_corr.sort_values()

plt.figure(figsize=(10, 7))

bars = plt.barh(
    target_corr.index,
    target_corr.values,
    color=[
        "steelblue" if value < 0 else "darkorange"
        for value in target_corr.values
    ]
)

# Add a reference line at zero
plt.axvline(0, color="black", linewidth=1)

plt.title(
    "Feature Correlations with Disease Progression",
    fontsize=15,
    fontweight="bold"
)
plt.xlabel("Pearson Correlation with Target")
plt.ylabel("Feature")

# Label each bar with its correlation
for bar, value in zip(bars, target_corr.values):
    offset = 0.015 if value >= 0 else -0.015
    plt.text(
        value + offset,
        bar.get_y() + bar.get_height() / 2,
        f"{value:.2f}",
        va="center",
        ha="left" if value >= 0 else "right",
        fontsize=9
    )

plt.xlim(min(target_corr.min() - 0.12, -0.1),
         max(target_corr.max() + 0.12, 0.1))

plt.tight_layout()
plt.show()
```

```python

# Visualization 6: Compare age, BMI, and disease progression

plt.figure(figsize=(11, 7))

scatter = plt.scatter(
    df["age"],
    df["bmi"],
    c=df["target"],
    cmap="viridis",
    s=75,
    alpha=0.8,
    edgecolors="white",
    linewidths=0.4
)

cbar = plt.colorbar(scatter)
cbar.set_label("Disease Progression Score", fontsize=11)

plt.title(
    "Age, BMI, and Disease Progression",
    fontsize=16,
    fontweight="bold"
)
plt.xlabel("Age (Standardized Feature Value)", fontsize=12)
plt.ylabel("BMI (Standardized Feature Value)", fontsize=12)

plt.tight_layout()
plt.show()
```


## Key Findings and Data Story

### 1. Distribution of Disease Progression
The histogram shows how disease-progression scores vary across the dataset. The mean is approximately 152.13, while the median is 140.5, providing two measures of the typical value.

### 2. Relationships Between Variables
The correlation heatmap reveals the direction and strength of linear relationships among the dataset's features. BMI has the strongest positive correlation with the target, at approximately 0.586.

### 3. BMI and Disease Progression
The scatter plot and trend line show that higher standardized BMI values tend to be associated with higher target values. However, the observations vary considerably around the trend line.

### 4. Comparison Across BMI Groups
The box plot compares target distributions for lower, middle, and higher BMI groups. Differences in medians and spreads help show how the groups vary.

### 5. Feature Correlation Ranking
The ranked bar chart makes it easier to compare features by their correlation with the target and identify relationships worth further investigation.

### 6. Exploring Multiple Variables
The multivariable scatter plot compares age and BMI, using color to represent target values. It provides a broader view of how the variables may relate to one another.

## Scientific Implications and Limitations

These findings can help identify patterns for further scientific investigation. BMI's positive association with the target may be worth studying alongside other variables. However, correlation does not prove causation, and these visualizations alone cannot establish clinical conclusions.

The dataset's target represents a quantitative measure of disease progression one year after baseline. It is not a diagnosis of diabetes. The dataset is also limited in size and uses standardized feature values, which should be considered when interpreting the results.
